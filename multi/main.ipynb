
#### 训练和验证命令
#### 训练

python -m multi.train_czsl --config multi/config.yaml
##### 可选参数：

--resume checkpoints/czsl_latest_checkpoint.pt - 恢复训练
零样本评估

##### 基础评估
python -m multi.evaluate_czsl --checkpoint checkpoints/czsl_best_model.pt --mode zero_shot --split test

##### 带可视化的评估
python -m multi.evaluate_czsl --checkpoint checkpoints/czsl_best_model.pt --mode zero_shot --split test --visualize --output_dir results

#### 参数说明：

--checkpoint: 模型检查点路径
--mode: zero_shot 或 supervised
--split: train, val 或 test
--visualize: 生成 t-SNE、混淆矩阵等可视化
--output_dir: 结果输出目录
